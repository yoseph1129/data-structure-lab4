```python
A = {
    0: [1, 2, 3],
    1: [0, 2, 4, 5],
    2: [0, 1, 6],
    3: [0],
    4: [1],
    5: [1],
    6: [2]
}

def bfs(A):
    # 방문 순서를 기록할 결과 리스트와 중복 방문 방지용 집합(Set) 초기화
    from collections import deque
    answer = []
    visited = set()

    # 시작 노드(0)를 큐에 삽입하고 방문 처리
    queue = deque([0])
    visited.add(0)

    # 큐가 빌 때까지 반복하여 탐색 진행
    while queue:
        # 큐의 맨 앞 노드를 꺼내어 결과 리스트에 기록
        node = queue.popleft()
        answer.append(node)

        # 번호가 작은 노드부터 방문하기 위해 인접 노드를 오름차순 정렬하여 순회
        for neighbor in sorted(A[node]):
            # 아직 방문하지 않은 이웃 노드가 있다면 큐에 삽입 후 방문 처리
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return answer


def dfs(A):
    # 방문 순서를 기록할 결과 리스트와 중복 방문 방지용 집합(Set) 초기화
    answer = []
    visited = set()

    # 시작 노드(0)를 스택에 삽입하고 방문 처리
    stack = [0]
    visited.add(0)

    # 스택이 빌 때까지 반복하여 탐색 진행
    while stack:
        # 스택의 맨 위(가장 최근 삽입된) 노드를 꺼내어 결과 리스트에 기록
        node = stack.pop()
        answer.append(node)

        # 후입선출(LIFO) 특성상 작은 번호가 먼저 나오도록 인접 노드를 내림차순 정렬하여 순회
        for neighbor in sorted(A[node], reverse=True):
            # 아직 방문하지 않은 이웃 노드가 있다면 스택에 삽입 후 방문 처리
            if neighbor not in visited:
                visited.add(neighbor)
                stack.append(neighbor)

    return answer

# 아래는 체크함수입니다. 수정하실 필요 없습니다.
bfs_result = bfs(A)
dfs_result = dfs(A)

assert bfs_result == [0,1,2,3,4,5,6]
assert dfs_result == [0,1,4,5,2,6,3]
print('PASSED!')

# 각 알고리즘별 탐색 순서를 직관적으로 확인할 수 있도록 화살표 기호로 연결하여 출력
print(f"BFS 탐색 순서: {' -> '.join(map(str, bfs_result))}")
print(f"DFS 탐색 순서: {' -> '.join(map(str, dfs_result))}")
```
